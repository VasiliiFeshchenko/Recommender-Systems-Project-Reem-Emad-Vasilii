```python
# Read the dataset location from the configuration file
import json
PROJECT_DIR = Path.cwd().resolve()

if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

with open(PROJECT_DIR / "config.json", "r") as f:
    config = json.load(f)


DATA_DIR = (PROJECT_DIR / config["data_dir"]).resolve()

print("Dataset directory:", DATA_DIR)
print("Dataset exists:", DATA_DIR.is_dir())
```

```python
# we start loading moveileens data and start with Ratings into pd dataframes

def load_ratings(data_dir):
    ratings = pd.read_csv(
        data_dir / "ratings.dat",
        sep="::",
        engine="python",
        header=None,
        names=["UserID", "MovieID", "Rating", "Timestamp"]
    )
    return ratings
```

```python


ratings = load_ratings(DATA_DIR)
```

```python
# we check first five ratings and see their shape

display(ratings)
```

### Temporal train/validation/test split

```python

# Sort ratings chronologically using NumPy's working sort order
timestamps = ratings["Timestamp"].to_numpy(copy=True)
order = np.argsort(timestamps, kind="stable")
ratings_sorted = ratings.iloc[order].reset_index(drop=True)

# Split: 70% train, 10% validation, 20% test
n = len(ratings_sorted)
train_end = int(n * 0.70)
val_end = int(n * 0.80)

train = ratings_sorted.iloc[:train_end].copy()
validation = ratings_sorted.iloc[train_end:val_end].copy()
test = ratings_sorted.iloc[val_end:].copy()

# Check split sizes
print("Total:", len(ratings_sorted))
print("Train:", len(train))
print("Validation:", len(validation))
print("Test:", len(test))
```
